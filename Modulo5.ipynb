> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Gerando os Dados e Criando o DataFrame</b></span>

```python
# GERANDO OS DADOS
np.random.seed(10)
n = 72

tempo = pd.date_range("2026-09-26 00:00", periods=n, freq="1h")

tendencia = np.linspace(25, 30, n)
sazonalidade = 3 * np.sin(2 * np.pi * np.arange(n) / 24)
ruido = np.random.normal(0, 0.3, n)

temperatura = tendencia + sazonalidade + ruido
temperatura[35] += 7
temperatura[52] -= 6

df_t = pd.DataFrame({
    "tempo": tempo,
    "temperatura": temperatura
})

# VISUALIZANDO OS DADOS
df_t.head()
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Criando e Visualizando os Gráficos</b></span>

```python
# GRÁFICO
fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(
    df_t["tempo"],
    df_t["temperatura"],
    linewidth=2,
    label="Temperatura"
)

ax.plot(
    df_t["tempo"],
    tendencia,
    linestyle="--",
    linewidth=2,
    label="Tendência"
)

ax.scatter(
    df_t["tempo"][35],
    df_t["temperatura"][35],
    s=90,
    zorder=5,
    label="Pico"
)

ax.scatter(
    df_t["tempo"][52],
    df_t["temperatura"][52],
    s=90,
    zorder=5,
    label="Queda"
)

ax.axvspan(
   df_t["tempo"][34],
    df_t["tempo"][36],
    alpha=0.2,
    label="Evento operacional"
)

ymin = df_t["temperatura"].min() - 3
ymax = df_t["temperatura"].max() + 3
ax.set_ylim(ymin, ymax)

# ANOTAÇÕES
caixa = dict(
    boxstyle="round,pad=0.3",
    fc="white",
    ec="gray",
    alpha=0.95
)

ax.annotate(
    "Eixo temporal",
    xy=(df_t["tempo"][68], ymin + 0.5),
    xytext=(df_t["tempo"][58], ymin + 2.0),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

ax.annotate(
    "Tendência\n(crescimento geral)",
    xy=(df_t["tempo"][60], tendencia[60]),
    xytext=(df_t["tempo"][50], ymax - 1.0),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

ax.annotate(
    "Sazonalidade\n(padrão repetitivo)",
    xy=(df_t["tempo"][18], df_t["temperatura"][18]),
    xytext=(df_t["tempo"][4], ymax - 1.0),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

ax.annotate(
    "Pico",
    xy=(df_t["tempo"][35], df_t["temperatura"][35]),
    xytext=(df_t["tempo"][28], ymax - 0.3),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

ax.annotate(
    "Queda",
    xy=(df_t["tempo"][52], df_t["temperatura"][52]),
    xytext=(df_t["tempo"][56], ymin + 3.0),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

ax.annotate(
    "Evento operacional",
    xy=(df_t["tempo"][35], ymin + 4),
    xytext=(df_t["tempo"][38], ymin + 1.5),
    arrowprops=dict(arrowstyle="->", lw=1.5),
    bbox=caixa,
    fontsize=10
)

# FORMATAÇÃO
ax.set_title("Série Temporal IoT - Elementos Principais")
ax.set_xlabel("Tempo")
ax.set_ylabel("Temperatura (°C)")
ax.grid(True)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# GERANDO OS DADOS
def gerar_dados(n=120):
    np.random.seed(10)

    tempo = pd.date_range(
        "2026-09-26 08:00",
        periods=n,
        freq="1min"
    )

    t = np.arange(n)

    sensor = (
        30
        + 0.03 * t
        + 2.5 * np.sin(2 * np.pi * t / 30)
        + np.random.normal(0, 1.2, n)
    )

    df = pd.DataFrame({
        "tempo": tempo,
        "sensor": sensor
    }).set_index("tempo")

# CRIANDO NOVAS VARIÁVEIS
    df["media_5"] = df["sensor"].rolling(5).mean()
    df["media_20"] = df["sensor"].rolling(20).mean()

    return df

# CRIANDO OS GRÁFICOS
def plotar_medias(df):
    plt.figure(figsize=(12, 5))

    plt.plot(
        df.index,
        df["sensor"],
        alpha=0.45,
        label="Sinal original"
    )

    plt.plot(
        df.index,
        df["media_5"],
        linewidth=2,
        label="Média móvel - 5 min"
    )

    plt.plot(
        df.index,
        df["media_20"],
        linewidth=2.5,
        label="Média móvel - 20 min"
    )

    plt.title("Média Móvel: Janela Curta x Janela Longa")
    plt.xlabel("Tempo")
    plt.ylabel("Valor do sensor")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

df = gerar_dados()
plotar_medias(df)
```

```python
def gerar_dados(n=120):
    np.random.seed(10)

    tempo = pd.date_range(
        "2026-09-26 08:00",
        periods=n,
        freq="1h"
    )

    t = np.arange(n)

    valor = (
        50
        + 0.08 * t
        + 5 * np.sin(2 * np.pi * t / 24)
        + np.random.normal(0, 1.2, n)
    )

    return pd.DataFrame({
        "tempo": tempo,
        "valor": valor
    }).set_index("tempo")


def metricas(real, previsto):
    mae = mean_absolute_error(real, previsto)
    rmse = np.sqrt(mean_squared_error(real, previsto))
    mape = np.mean(np.abs((real - previsto) / real)) * 100

    return mae, rmse, mape


def avaliar_modelos(teste):
    modelos = ["naive", "seasonal_naive", "media_movel_6h"]
    resultados = []

    for modelo in modelos:
        mae, rmse, mape = metricas(
            teste["valor"],
            teste[modelo]
        )

        resultados.append({
            "Modelo": modelo,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE (%)": mape
        })

    return pd.DataFrame(resultados).sort_values("RMSE")


def plotar_treino_teste(treino, teste):
    plt.figure(figsize=(12, 4))

    plt.plot(treino.index, treino["valor"], label="Treino")
    plt.plot(teste.index, teste["valor"], label="Teste")
    plt.axvline(teste.index[0], linestyle="--")

    plt.title("Treino x Teste")
    plt.xlabel("Tempo")
    plt.ylabel("Valor")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.show()


def plotar_previsoes(teste):
    plt.figure(figsize=(12, 5))

    plt.plot(
        teste.index,
        teste["valor"],
        marker="o",
        label="Real"
    )

    plt.plot(
        teste.index,
        teste["naive"],
        linestyle="--",
        label="Naive"
    )

    plt.plot(
        teste.index,
        teste["seasonal_naive"],
        linestyle="--",
        label="Seasonal Naive"
    )

    plt.plot(
        teste.index,
        teste["media_movel_6h"],
        linestyle="--",
        label="Média móvel 6h"
    )

    plt.title("Baseline de Previsão")
    plt.xlabel("Tempo")
    plt.ylabel("Valor")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.show()


def plotar_erros(teste):
    erros = {
        "Erro Naive": np.abs(teste["valor"] - teste["naive"]),
        "Erro Seasonal Naive": np.abs(
            teste["valor"] - teste["seasonal_naive"]
        ),
        "Erro Média móvel": np.abs(
            teste["valor"] - teste["media_movel_6h"]
        )
    }

    plt.figure(figsize=(12, 4))

    for nome, erro in erros.items():
        plt.plot(teste.index, erro, label=nome)

    plt.title("Erro Absoluto")
    plt.xlabel("Tempo")
    plt.ylabel("Erro")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.show()


# GERANDO OS DADOS
df = gerar_dados()

# TREINO E TESTE
h = 24

treino = df.iloc[:-h].copy()
teste = df.iloc[-h:].copy()

# MODELO BASELINES
teste["naive"] = treino["valor"].iloc[-1]

teste["seasonal_naive"] = (
    treino["valor"]
    .iloc[-24:]
    .values
)

teste["media_movel_6h"] = (
    treino["valor"]
    .tail(6)
    .mean()
)

# MÉTRICAS
resultado = avaliar_modelos(teste)
print(resultado.round(3))

# GRÁFICOS
plotar_treino_teste(treino, teste)
plotar_previsoes(teste)
plotar_erros(teste)
```

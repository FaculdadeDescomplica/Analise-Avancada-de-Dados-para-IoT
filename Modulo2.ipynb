```python
# Exemplo: residência inteligente com consumo por cômodo
horas = np.arange(24)
comodos = ['Sala','Quarto','Cozinha','Lavanderia']
linhas=[]
for c in comodos:
    base = {'Sala':1.2,'Quarto':0.8,'Cozinha':1.5,'Lavanderia':1.0}[c]
    for h in horas:
        consumo = base + 0.4*np.sin((h-18)/24*2*np.pi) + np.random.normal(0,.12)
        linhas.append([h,c,max(consumo,0.1)])
df = pd.DataFrame(linhas, columns=['hora','comodo','kwh'])
resumo = df.groupby('comodo')['kwh'].mean().sort_values(ascending=False)
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Visualizando o Resumo</b></span>

```python
print(resumo)
```

```python
resumo.plot(kind='bar', figsize=(8,4), title='Consumo médio por cômodo')
plt.ylabel('kWh médio')
plt.grid(axis='y', alpha=.3)
plt.show()
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Gerando os Dados e Criando o DataFrame</b></span>

```python
# GERARANDO DADOS IoT
np.random.seed(42)

tempo = pd.date_range("2026-01-01", periods=240, freq="h")
hora = tempo.hour.to_numpy()

temperatura = (
    24
    + 3 * np.sin(2 * np.pi * hora / 24)
    + np.random.normal(0, 0.8, 240)
)

energia = (
    5
    + 0.22 * temperatura
    + 1.2 * np.sin(2 * np.pi * (hora - 3) / 24)
    + np.random.normal(0, 0.6, 240)
)

# CRIAR DATAFRAME
df = pd.DataFrame({
    "tempo": tempo,
    "temperatura": temperatura,
    "energia_kwh": energia
}).set_index("tempo")
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Visualizando os Dados</b></span>

```python
styled_df = df.head().style

# Definindo a cor de fundos, a cor do texto, e a cor do DataFrame
styled_df.set_properties(**{"background-color": "#254E58", "color": "#e9c46a", "border": "1.5px solid black"})

# Modificando a cor de fundo dos cabeçalhos da tabela
styled_df.set_table_styles([
    {"selector": "th", "props": [("color", 'white'), ("background-color", "#333333")]}
])
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Criando Análises e Visualizando os Resultados</b></span>

```python
# 3. ANÁLISE
df["energia_media_6h"] = df["energia_kwh"].rolling(6).mean()

correlacao = df["temperatura"].corr(df["energia_kwh"])

media = df["energia_kwh"].mean()
desvio = df["energia_kwh"].std()
limite = media + 1.5 * desvio

df["alerta"] = df["energia_kwh"] > limite

# 4. RESULTADOS
print(df.head())

print(f"\nCorrelação: {correlacao:.2f}")
print(f"Consumo médio: {media:.2f} kWh")
print(f"Limite de alerta: {limite:.2f} kWh")
print(f"Quantidade de alertas: {df['alerta'].sum()}")
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Criando e Visualizando os Gráficos</b></span>

```python
# GRÁFICO TEMPORAL
plt.figure(figsize=(11, 5))

plt.plot(df.index, df["temperatura"], label="Temperatura")
plt.plot(df.index, df["energia_kwh"], label="Energia")
plt.plot(df.index, df["energia_media_6h"], label="Média móvel 6h")

plt.title("Leituras IoT ao Longo do Tempo")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

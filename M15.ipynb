```python
def gerar_dados():
    np.random.seed(42)

    df = pd.DataFrame({
        "tempo": pd.date_range("2026-10-05", periods=48, freq="1h"),
        "temperatura": 58 + np.random.normal(0, 1.2, 48),
        "vibracao": 2.0 + np.random.normal(0, 0.18, 48)
    })

    df.loc[30:, "temperatura"] += np.linspace(2, 10, 18)
    df.loc[30:, "vibracao"] += np.linspace(0.3, 2.0, 18)

    return df


df = gerar_dados()

# 1. RISCO
df["score_risco"] = (
    np.maximum(df["temperatura"] - 65, 0) * 2
    + np.maximum(df["vibracao"] - 3, 0) * 10
)

df["status"] = pd.cut(
    df["score_risco"],
    [-1, 1, 8, np.inf],
    labels=["Normal", "Atenção", "Manutenção"]
)

# 2. SENSORES
plt.figure(figsize=(10, 4))
plt.plot(df["tempo"], df["temperatura"], color="blue", linewidth=2, label="Temperatura (°C)")
plt.plot(df["tempo"], df["vibracao"] * 20, color="orange", linewidth=2, label="Vibração (escala visual)")
plt.title("Evolução dos Sensores da Máquina")
plt.xlabel("Tempo")
plt.ylabel("Nível do sinal")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 3. RISCO
plt.figure(figsize=(10, 4))
plt.plot(df["tempo"], df["score_risco"], color="black", linewidth=2, label="Risco")
plt.axhline(1, color="orange", linestyle="--", linewidth=2, label="Atenção")
plt.axhline(8, color="red", linestyle="--", linewidth=2, label="Manutenção")
plt.title("Índice de Risco para Manutenção Inteligente")
plt.xlabel("Tempo")
plt.ylabel("Score de risco")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 4. STATUS
status = df["status"].value_counts().reindex(
    ["Normal", "Atenção", "Manutenção"], fill_value=0
)

plt.figure(figsize=(7, 4))
barras = plt.bar(status.index, status.values, color=["green", "orange", "red"])
plt.bar_label(barras, padding=3)
plt.title("Status Operacional da Máquina")
plt.ylabel("Quantidade de leituras")
plt.grid(axis="y", alpha=0.3)
plt.show()

# 5. RESULTADOS
print(df.tail(12).round(2))
```

```python
def gerar_dados(n=24):
    np.random.seed(42)

    df = pd.DataFrame({
        "tempo": pd.date_range("2026-10-05 08:00", periods=n, freq="30min"),
        "velocidade_kmh": 55 + np.random.normal(0, 7, n),
        "temperatura_c": 4 + np.random.normal(0, 0.35, n),
        "combustivel_pct": np.linspace(85, 48, n) + np.random.normal(0, 1.2, n),
        "latitude": np.linspace(-22.25, -22.40, n),
        "longitude": np.linspace(-45.70, -45.95, n)
    })

    df.loc[15:18, "temperatura_c"] += [1.2, 2.0, 2.7, 1.5]

    progresso = np.arange(n) / (n - 1)
    df["distancia_restante_km"] = 42 * (1 - progresso)
    df["eta_min"] = df["distancia_restante_km"] / np.clip(df["velocidade_kmh"], 20, None) * 60

    df["alerta_temp"] = df["temperatura_c"] > 5
    df["alerta_comb"] = df["combustivel_pct"] < 20

    return df


def grafico(x, y, titulo, ylabel, label=None, limite=None):
    plt.figure(figsize=(10, 4))
    plt.plot(x, y, linewidth=2, label=label)

    if limite:
        plt.axhline(limite[0], linestyle="--", label=limite[1])

    plt.title(titulo)
    plt.xlabel("Tempo")
    plt.ylabel(ylabel)
    plt.grid(alpha=0.3)

    if label or limite:
        plt.legend()

    plt.show()


df = gerar_dados()
idx_atual = 17
atual = df.iloc[idx_atual]
dados = df.iloc[:idx_atual + 1]

# 1. LOCALIZAÇÃO
plt.figure(figsize=(7, 5))
plt.plot(df["longitude"], df["latitude"], marker="o", label="Rota")
plt.scatter(atual["longitude"], atual["latitude"], s=140, label="Veículo agora")
plt.scatter(-45.95, -22.40, s=140, marker="X", label="Destino")
plt.title("Localização Atual do Veículo")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 2. VELOCIDADE
grafico(
    dados["tempo"], dados["velocidade_kmh"],
    "Velocidade do Veículo", "Velocidade (km/h)"
)

# 3. TEMPERATURA
grafico(
    dados["tempo"], dados["temperatura_c"],
    "Temperatura da Carga", "Temperatura (°C)",
    "Temperatura", (5, "Limite = 5 °C")
)

alertas = dados[dados["alerta_temp"]]
plt.figure(figsize=(10, 4))
plt.plot(dados["tempo"], dados["temperatura_c"], linewidth=2, label="Temperatura")
plt.axhline(5, linestyle="--", label="Limite = 5 °C")
plt.scatter(alertas["tempo"], alertas["temperatura_c"], s=70, label="Alerta")
plt.title("Temperatura da Carga")
plt.xlabel("Tempo")
plt.ylabel("Temperatura (°C)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 4. COMBUSTÍVEL
grafico(
    dados["tempo"], dados["combustivel_pct"],
    "Nível de Combustível", "Combustível (%)",
    limite=(20, "Limite = 20%")
)

# 5. DISTÂNCIA E ETA
plt.figure(figsize=(10, 4))
plt.plot(dados["tempo"], dados["distancia_restante_km"], linewidth=2, label="Distância restante (km)")
plt.plot(dados["tempo"], dados["eta_min"], linewidth=2, label="ETA (min)")
plt.title("Distância Restante e Tempo Estimado de Chegada")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 6. RESUMO
print("RESUMO ATUAL DO VEÍCULO")
print(f"Latitude: {atual['latitude']:.4f}")
print(f"Longitude: {atual['longitude']:.4f}")
print(f"Velocidade atual: {atual['velocidade_kmh']:.1f} km/h")
print(f"Distância restante: {atual['distancia_restante_km']:.1f} km")
print(f"Tempo estimado para chegada: {atual['eta_min']:.0f} min")
print(f"Temperatura da carga: {atual['temperatura_c']:.1f} °C")
print(f"Combustível: {atual['combustivel_pct']:.1f}%")
print(f"Alertas de temperatura: {int(dados['alerta_temp'].sum())}")
```

```python
def gerar_dados():
    np.random.seed(42)
    x = np.linspace(0, 2*np.pi, 24)

    df = pd.DataFrame({
        "tempo": pd.date_range("2026-10-05", periods=24, freq="1h"),
        "umidade_solo": 62 - np.linspace(0, 25, 24) + np.random.normal(0, 2, 24),
        "temperatura": 24 + 5*np.sin(x) + np.random.normal(0, 0.8, 24),
        "luminosidade": np.clip(
            900*np.sin(np.linspace(0, np.pi, 24)) + np.random.normal(0, 40, 24),
            0, None
        )
    })

    df["irrigar"] = df["umidade_solo"] < 40
    df["risco"] = np.maximum(40 - df["umidade_solo"], 0)*2 + \
                  np.maximum(df["temperatura"] - 30, 0)*3

    return df


def grafico(coluna, titulo, ylabel, cor, limite=None):
    plt.figure(figsize=(10, 4))
    plt.plot(df["tempo"], df[coluna], color=cor, linewidth=2)

    if limite:
        plt.axhline(limite[0], color="red", linestyle="--",
                    linewidth=2, label=limite[1])
        plt.legend()

    plt.title(titulo)
    plt.xlabel("Tempo")
    plt.ylabel(ylabel)
    plt.grid(alpha=0.3)
    plt.show()


df = gerar_dados()

# 1. UMIDADE
alertas = df[df["irrigar"]]

plt.figure(figsize=(10, 4))
plt.plot(df["tempo"], df["umidade_solo"], color="green", linewidth=2, label="Umidade do solo")
plt.axhline(40, color="red", linestyle="--", linewidth=2, label="Limite = 40%")
plt.scatter(alertas["tempo"], alertas["umidade_solo"], color="red", s=70, label="Necessidade de irrigação")
plt.title("Umidade do Solo ao Longo do Dia")
plt.xlabel("Tempo")
plt.ylabel("Umidade do solo (%)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 2. TEMPERATURA
grafico("temperatura", "Temperatura do Ambiente", "Temperatura (°C)",
        "orange", (30, "Limite = 30 °C"))

# 3. LUMINOSIDADE
grafico("luminosidade", "Luminosidade ao Longo do Dia",
        "Luminosidade (lux)", "goldenrod")

# 4. RISCO
grafico("risco", "Índice de Risco Agrícola", "Score de risco", "purple")

# 5. RESULTADOS
print("Umidade mínima:", round(df["umidade_solo"].min(), 1), "%")
print("Temperatura máxima:", round(df["temperatura"].max(), 1), "°C")
print("Horas com necessidade de irrigação:", int(df["irrigar"].sum()))
print("Maior score de risco:", round(df["risco"].max(), 1))
```

```python
def estilo(titulo, xlabel, ylabel):
    ax = plt.gca()
    ax.set_facecolor("#303030")
    ax.set_title(titulo, color="white", fontweight="bold")
    ax.set_xlabel(xlabel, color="white")
    ax.set_ylabel(ylabel, color="white")
    ax.tick_params(colors="white")

    for s in ax.spines.values():
        s.set_color("#777777")


# 1. UMIDADE DO SOLO
np.random.seed(7)

dias = np.arange(1, 31)
umidade = 60 + 20*np.sin(dias/5) + np.random.normal(0, 5, 30)

plt.figure(figsize=(8, 4), facecolor="#252525")
plt.plot(dias, umidade, "o-", color="#49B9AF", linewidth=2, label="Umidade do solo")
plt.axhline(40, color="#E46B5B", linestyle="--", label="Alerta Seca")
plt.axhline(80, color="#6AA2D8", linestyle="--", label="Saturação")

estilo("Monitoramento IoT — Umidade do Solo", "Dia do mês", "Umidade do solo (%)")
plt.grid(alpha=0.18, color="white")
plt.legend(facecolor="#303030", edgecolor="#777777", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.show()


# 2. PRODUTIVIDADE
anos = ["2019", "2020", "2021", "2022", "2023", "2024"]
produtividade = [3.6, 4.2, 5.1, 4.0, 4.8, 5.3]
cores = ["#5CC2C4", "#5CC2C4", "#E46B5B", "#5CC2C4", "#5CC2C4", "#E46B5B"]

plt.figure(figsize=(7, 4), facecolor="#252525")
plt.bar(anos, produtividade, color=cores, width=0.55)

estilo("Produtividade da Lavoura", "Ano", "Produtividade (t/ha)")
plt.grid(axis="y", alpha=0.15, color="white")
plt.tight_layout()
plt.show()
```

```python
def gerar_dados():
    np.random.seed(42)

    horas = pd.date_range("2026-10-05", periods=24, freq="1h")
    trafego = 250 + 180*np.sin(np.linspace(-1, 2*np.pi - 1, 24)) + np.random.normal(0, 35, 24)
    trafego = np.clip(trafego, 60, None)

    df = pd.DataFrame({
        "hora": horas,
        "veiculos_h": trafego,
        "pm25": 20 + 0.035*trafego + np.random.normal(0, 3, 24),
        "iluminacao_pct": np.clip(100 - 4*np.arange(24), 15, 100)
    })

    df["congestionamento"] = df["veiculos_h"] > 400
    df["alerta_ar"] = df["pm25"] > 35
    return df


def grafico(coluna, titulo, ylabel, cor, limite=None, alerta=None):
    plt.figure(figsize=(10, 4))
    plt.plot(df["hora"], df[coluna], color=cor, linewidth=2, label=ylabel)

    if limite:
        plt.axhline(limite[0], color=limite[1], linestyle="--", linewidth=2, label=limite[2])

    if alerta:
        dados = df[df[alerta[0]]]
        plt.scatter(dados["hora"], dados[coluna], color="red", s=70, label=alerta[1])

    plt.title(titulo)
    plt.xlabel("Tempo")
    plt.ylabel(ylabel)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()


df = gerar_dados()

# 1. TRÁFEGO
grafico(
    "veiculos_h", "Fluxo de Tráfego Urbano", "Veículos por hora", "blue",
    (400, "red", "Limite = 400 veículos/h"),
    ("congestionamento", "Congestionamento")
)

# 2. QUALIDADE DO AR
grafico(
    "pm25", "Qualidade do Ar - PM2.5", "PM2.5", "green",
    (35, "orange", "Limite didático = 35"),
    ("alerta_ar", "Alerta")
)

# 3. ILUMINAÇÃO
grafico(
    "iluminacao_pct", "Nível de Iluminação Pública",
    "Potência da iluminação (%)", "purple"
)

# 4. RESULTADOS
print("Pico de tráfego:", int(df["veiculos_h"].max()), "veículos/h")
print("Horas com congestionamento:", int(df["congestionamento"].sum()))
print("PM2.5 máximo:", round(df["pm25"].max(), 1))
print("Horas com alerta de qualidade do ar:", int(df["alerta_ar"].sum()))
```

```python
def gerar_dados():
    np.random.seed(42)

    horas = pd.date_range("2026-10-05", periods=24, freq="1h")
    consumo = 120 + 35*np.sin(np.linspace(-1.2, 2*np.pi - 1.2, 24)) + np.random.normal(0, 6, 24)
    consumo = np.clip(consumo, 70, None)
    consumo[17:20] += [30, 45, 35]

    df = pd.DataFrame({
        "hora": horas,
        "consumo_kwh": consumo,
        "tensao_v": 220 + np.random.normal(0, 4, 24)
    })

    df["alerta_consumo"] = df["consumo_kwh"] > 150
    df["alerta_tensao"] = (df["tensao_v"] < 210) | (df["tensao_v"] > 230)
    df["tarifa_rs_kwh"] = np.where(df["hora"].dt.hour.between(18, 21), 1.20, 0.75)
    df["custo_rs"] = df["consumo_kwh"] * df["tarifa_rs_kwh"]

    return df


df = gerar_dados()

# 1. CONSUMO
alertas = df[df["alerta_consumo"]]

plt.figure(figsize=(10, 4))
plt.plot(df["hora"], df["consumo_kwh"], color="blue", linewidth=2, label="Consumo")
plt.axhline(150, color="red", linestyle="--", linewidth=2, label="Limite = 150 kWh")
plt.scatter(alertas["hora"], alertas["consumo_kwh"], color="red", s=70, label="Pico de consumo")
plt.title("Consumo de Energia ao Longo do Dia")
plt.xlabel("Tempo")
plt.ylabel("Consumo (kWh)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 2. TENSÃO
plt.figure(figsize=(10, 4))
plt.plot(df["hora"], df["tensao_v"], color="green", linewidth=2, label="Tensão")
plt.axhline(210, color="orange", linestyle="--", label="Limite inferior")
plt.axhline(230, color="red", linestyle="--", label="Limite superior")
plt.title("Qualidade da Energia - Tensão")
plt.xlabel("Tempo")
plt.ylabel("Tensão (V)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 3. CUSTO
plt.figure(figsize=(10, 4))
plt.bar(df["hora"].dt.hour.astype(str), df["custo_rs"], color="purple")
plt.title("Custo de Energia por Hora")
plt.xlabel("Hora")
plt.ylabel("Custo (R$)")
plt.grid(axis="y", alpha=0.3)
plt.show()

# 4. RESULTADOS
print("Consumo total:", round(df["consumo_kwh"].sum(), 1), "kWh")
print("Pico de consumo:", round(df["consumo_kwh"].max(), 1), "kWh")
print("Horas com pico:", int(df["alerta_consumo"].sum()))
print(f"Custo total: R$ {df['custo_rs'].sum():.2f}")
print("Alertas de tensão:", int(df["alerta_tensao"].sum()))
```

```python
# GERANDO OS DADOS
np.random.seed(42)
n = 160

df = pd.DataFrame({
    "batimentos": np.random.normal(78, 10, n),
    "saturacao": np.random.normal(96, 2, n),
    "temperatura": np.random.normal(36.7, 0.5, n),
    "sono_horas": np.random.normal(6.5, 1.1, n)
})
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Trantando os Dados</b></span>

```python
# TRATAR OS DADOS
indices_nulos = np.random.choice(df.index, 8, replace=False)
df.loc[indices_nulos, "saturacao"] = np.nan

df = df.drop_duplicates()
df["saturacao"] = df["saturacao"].fillna(df["saturacao"].median())

# DEFINIR RISCO
df["risco_alto"] = (
    (df["batimentos"] > 92) |
    (df["saturacao"] < 94) |
    (df["temperatura"] > 37.5)
).astype(int)
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Criando a Modelagem</b></span>

```python
# TREINO E TESTE
X = df[["batimentos", "saturacao", "temperatura", "sono_horas"]]
y = df["risco_alto"]

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# TREINAR O MODELO
modelo = LogisticRegression(max_iter=1000)
modelo.fit(X_treino, y_treino)

predicoes = modelo.predict(X_teste)
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Avaliando a Performance do Modelo</b></span>

```python
# AVALIAR MODELO
acuracia = accuracy_score(y_teste, predicoes)
matriz = confusion_matrix(y_teste, predicoes)

print(f"\nAcurácia: {acuracia:.3f}")
print("\nMatriz de confusão:")
print(matriz)
```

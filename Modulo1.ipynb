```python
# BIBLIOTECAS
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# CONFIGURAÇÕES DE ESTÉTICA👇🏽

# Ajustes visuais do Jupyter
from IPython.core.display import HTML
display(HTML('<style>.container { width:80% !important; }</style>'))
# pd.set_option('display.max_columns', 60)

# Configurando pandas para mostrar todas as linhas e colunas
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None) 

# Configurando pandas para não mostrar notação científica para números
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Serve para Manter os gráficos nesse mesmo Notebook
%matplotlib inline
# Ignorando alertas
import warnings
warnings.filterwarnings("ignore")

# Reprodutibilidade
np.random.seed(42)
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Gerando os Dados e Criando o DataFrame</b></span>

```python

# GERANDO OS DADOS
horas = np.arange(24)
fluxo = np.random.poisson(90, 24) + (horas >= 18) * 30
tempo = np.random.normal(7, 1.2, 24)

vendas = (
    fluxo * (0.06 + tempo / 250) +
    np.random.normal(0, 2, 24)
).round().clip(0)

# CRIAndo DATAFRAME
df = pd.DataFrame({
    "hora": horas,
    "fluxo_pessoas": fluxo,
    "tempo_medio_min": tempo.round(1),
    "vendas": vendas
})
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Visualizando os Dados</b></span>

```python
styled_df = df.head().style

# Definindo a cor de fundos, a cor do texto, e a cor do DataFrame
styled_df.set_properties(**{"background-color": "#254E58", "color": "#e9c46a", "border": "1.5px solid black"})

# Modificando a cor de fundo dos cabeçalhos da tabela
styled_df.set_table_styles([
    {"selector": "th", "props": [("color", 'white'), ("background-color", "#333333")]}
])
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Calculando a Conversão média</b></span>

```python
df["conversao"] = df["vendas"] / df["fluxo_pessoas"]
print(f"\nConversão média: {df['conversao'].mean():.3f}")
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Criando e Visualizando os Gráficos</b></span>

```python
# GRÁFICO
plt.figure(figsize=(9, 4))
plt.plot(df["hora"], df["conversao"], marker="o")

plt.title("Conversão por Hora - Ambiente IoT")
plt.xlabel("Hora")
plt.ylabel("Taxa de Conversão")
plt.grid(alpha=0.3)
plt.show()
```

> <span style='font-size:15px; font-family:Verdana;color: #c6dadfff; border-bottom: solid 3px'><b>Verificando o melhor e o Pior Horário</b></span>

```python
# MELHOR E PIOR HORÁRIO
melhor = df.loc[df["conversao"].idxmax()]
pior = df.loc[df["conversao"].idxmin()]

print(f"Melhor horário: {melhor['hora']:.0f}h → Conversão: {melhor['conversao']:.3f}")
print(f"Pior horário: {pior['hora']:.0f}h → Conversão: {pior['conversao']:.3f}")
```

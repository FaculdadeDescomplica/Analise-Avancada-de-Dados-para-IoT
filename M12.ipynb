```python
import pandas as pd
import matplotlib.pyplot as plt


def grafico(coluna, titulo, ylabel, divisor=1):
    plt.figure(figsize=(8, 4))
    plt.bar(dados["sensor"], dados[coluna] / divisor)
    plt.title(titulo)
    plt.xlabel("Tipo de sensor")
    plt.ylabel(ylabel)
    plt.grid(axis="y", alpha=0.3)
    plt.show()


# 1. DADOS
dados = pd.DataFrame({
    "sensor": ["Temperatura", "Vibração", "Pressão", "Energia"],
    "quantidade": [500, 300, 200, 100],
    "leituras_por_segundo": [1, 5, 2, 1]
})

dados["msg_por_seg"] = dados["quantidade"] * dados["leituras_por_segundo"]
dados["msg_por_dia"] = dados["msg_por_seg"] * 86400

# 2. GRÁFICOS
grafico(
    "msg_por_seg",
    "Velocidade: Mensagens Geradas por Segundo",
    "Mensagens por segundo"
)

grafico(
    "msg_por_dia",
    "Volume: Mensagens Geradas por Dia",
    "Milhões de mensagens/dia",
    1_000_000
)

# 3. RESULTADOS
print("Mensagens por segundo:", dados["msg_por_seg"].sum())
print("Mensagens por dia:", dados["msg_por_dia"].sum())
print(dados)
```

**Conclusão**
* IoT pode gerar dados em grande volume, alta velocidade e grande variedade. Quando essa escala exige tecnologias específicas para armazenamento, processamento e análise, temos um cenário de Big Data.

<h2 style="font-family: Verdana; font-size: 25px; font-style: normal; font-weight: normal; text-decoration: none; text-transform: none; letter-spacing: 2px; color: #254E58; background-color: #ffffff;"><b> Vídeo</b> 3

```python
# =========================================
# 1. BRONZE - DADOS BRUTOS
# =========================================

bronze = pd.DataFrame({
    "sensor_id": [
        "S1", "S1", "S2", "S2",
        "S3", "S3", "S1", "S2"
    ],

    "temperatura": [
        30.2, 31.1, 29.8, np.nan,
        35.5, 120.0, 30.9, 29.5
    ],

    "umidade": [
        60, 59, 65, 64,
        55, 54, 58, 66
    ],

    "timestamp": pd.date_range(
        "2026-10-02 08:00",
        periods=8,
        freq="5min"
    )
})

# Duplicidade
bronze = pd.concat(
    [
        bronze,
        bronze.iloc[[1]]
    ],
    ignore_index=True
)

# =========================================
# 2. SILVER - DADOS LIMPOS
# =========================================

silver = (
    bronze
    .drop_duplicates()
    .dropna()
    .query("temperatura < 80")
    .copy()
)

# =========================================
# 3. GOLD - DADOS ANALÍTICOS
# =========================================

gold = (
    silver
    .groupby("sensor_id")
    .agg(
        temp_media=(
            "temperatura",
            "mean"
        ),

        umidade_media=(
            "umidade",
            "mean"
        ),

        leituras=(
            "sensor_id",
            "size"
        )
    )
    .reset_index()
)

# =========================================
# 4. GRÁFICO - VOLUME
# =========================================

plt.figure(figsize=(7, 4))

plt.bar(
    ["Bronze", "Silver", "Gold"],
    [
        len(bronze),
        len(silver),
        len(gold)
    ]
)

plt.title(
    "Volume de Dados por Camada"
)

plt.ylabel(
    "Quantidade de registros"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# =========================================
# 5. GRÁFICO - TEMPERATURA
# =========================================

plt.figure(figsize=(7, 4))

plt.bar(
    gold["sensor_id"],
    gold["temp_media"]
)

plt.title(
    "Gold: Temperatura Média por Sensor"
)

plt.xlabel("Sensor")
plt.ylabel("Temperatura média (°C)")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# =========================================
# 6. GRÁFICO - UMIDADE
# =========================================

plt.figure(figsize=(7, 4))

plt.bar(
    gold["sensor_id"],
    gold["umidade_media"]
)

plt.title(
    "Gold: Umidade Média por Sensor"
)

plt.xlabel("Sensor")
plt.ylabel("Umidade média (%)")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# =========================================
# 7. RESULTADOS
# =========================================

print("BRONZE")
print(bronze)

print("\nSILVER")
print(silver)

print("\nGOLD")
print(
    gold.round(2)
)
```

```python
def gerar_dados():
    np.random.seed(42)

    maquinas = ["M1", "M2", "M3"]
    minutos = pd.date_range("2026-10-02 08:00", periods=60, freq="1min")
    bases = {"M1": 60, "M2": 65, "M3": 70}

    dados = [
        [m, t, bases[m] + np.random.normal(0, 1.5)]
        for m in maquinas
        for t in minutos
        for _ in range(20)
    ]

    return pd.DataFrame(dados, columns=["maquina", "tempo", "temperatura"])


df = gerar_dados()
maquinas = ["M1", "M2", "M3"]

agregado = (
    df.groupby(["maquina", "tempo"])["temperatura"]
    .mean()
    .reset_index()
)

# 1. VOLUME
plt.figure(figsize=(7, 4))
plt.bar(["Leituras brutas", "Após agregação"], [len(df), len(agregado)])
plt.title("Redução do Volume após Processamento")
plt.ylabel("Quantidade de registros")
plt.grid(axis="y", alpha=0.3)
plt.show()

# 2. TEMPERATURA MÉDIA
plt.figure(figsize=(10, 4))

for maquina in maquinas:
    parte = agregado[agregado["maquina"] == maquina]
    plt.plot(parte["tempo"], parte["temperatura"], linewidth=2, label=maquina)

plt.title("Temperatura Média por Máquina")
plt.xlabel("Tempo")
plt.ylabel("Temperatura média (°C)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
